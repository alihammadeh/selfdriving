```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from glob import glob
import IPython.display as display
import json
import os
from google.colab import drive
drive.mount('/content/drive')
drive_folder_path = '/content/drive/My Drive/selfdriving'
resized_image_path = os.path.join(drive_folder_path, 'resized_images')
```

```python
# Randomly split the data
#  -> 70 - 20 - 10
#  -> range(len(images))
#  -> randomize the range list
#  -> split into 70 - 20 - 10 lists
#  -> for each iid in train:
#  -> copy image into train file
#  -> copy annos into text file
#  ->
annotations_path = os.path.join(resized_image_path, 'resized_annos_id.json')
annos_file = open(annotations_path, 'r')
annos_json = json.load(annos_file)
annos_file.close()
```

```python
image_path_list = os.path.join(drive_folder_path, 'export','_annotations.coco.json')
image_path_file = open(image_path_list, 'r')
image_path_json = json.load(image_path_file)
image_path_file.close()
```

```python
image_paths = [image['file_name'] for image in image_path_json['images']]
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

images_train, images_test = train_test_split(image_paths, test_size=0.3, random_state=42)
```

```python
images_dev, images_test = train_test_split(images_test, test_size=0.5, random_state=42)
```

```python
train_path = os.path.join(resized_image_path, 'train')
dev_path = os.path.join(resized_image_path, 'dev')
test_path = os.path.join(resized_image_path, 'test')

if not os.path.exists(train_path):
  os.makedirs(train_path)

if not os.path.exists(dev_path):
  os.makedirs(dev_path)

if not os.path.exists(test_path):
  os.makedirs(test_path)

for path in images_train:
  abs_path = os.path.join(resized_image_path, path)
  img = Image.open(abs_path)
  img.save(os.path.join(train_path, path))
  os.remove(abs_path)

for path in images_test:
  abs_path = os.path.join(resized_image_path, path)
  img = Image.open(abs_path)
  img.save(os.path.join(test_path, path))
  os.remove(abs_path)

for path in images_dev:
  abs_path = os.path.join(resized_image_path, path)
  img = Image.open(abs_path)
  img.save(os.path.join(dev_path, path))
  os.remove(abs_path)
```
